# Validating with the Schema

Validating an SDC XML file against an XML Schema (XSD) checks that the document follows the expected structure, datatypes, and constraints. This step catches structural errors early and gives you precise line/column feedback for fixes.

## What you need
- **Your XML file** – e.g., an SDC form or an instance document produced from a form.
- **The XSD schema(s)** – SDC often relies on one or more XSD files. Sometimes a “root” schema imports others. Keep the full set together in the same folder when validating locally.
- (Optional) An **`xsi:schemaLocation`** hint in your XML. This is helpful, but not required if your validator lets you point to the schema files directly.

> Tip: If the XML uses XML namespaces, the schema must match those namespaces. Mismatched target namespaces are a common reason for validation failures.

## Ways to validate
Choose whichever fits your workflow. Below are examples that work well in CI/CD and local development.

### 1) Command line: `xmllint` (libxml2)
```bash
# Validate with a single XSD
xmllint --noout --schema schema.xsd my-document.xml

# Validate when the XSD imports/includes other XSDs (keep them in the same folder)
xmllint --noout --schema sdc-root.xsd my-document.xml
```
- **Success:** No output and exit code `0`.
- **Failure:** Error messages with line/column numbers and a non‑zero exit code.

### 2) Java (Xerces/JAXP)
```java
import java.io.File;
import javax.xml.XMLConstants;
import javax.xml.transform.stream.StreamSource;
import javax.xml.validation.Schema;
import javax.xml.validation.SchemaFactory;
import javax.xml.validation.Validator;

public class ValidateXml {
  public static void main(String[] args) throws Exception {
    File xsd = new File("sdc-root.xsd");
    File xml = new File("my-document.xml");

    SchemaFactory sf = SchemaFactory.newInstance(XMLConstants.W3C_XML_SCHEMA_NS_URI);
    Schema schema = sf.newSchema(xsd);
    Validator validator = schema.newValidator();
    validator.validate(new StreamSource(xml));
    System.out.println("Valid.");
  }
}
```
- Wrap the `validate` call in a try/catch to print readable error messages for build logs.

### 3) Python (`lxml`)
```python
from lxml import etree

with open("sdc-root.xsd", "rb") as f:
    schema_doc = etree.parse(f)
schema = etree.XMLSchema(schema_doc)

xml_doc = etree.parse("my-document.xml")
schema.assertValid(xml_doc)     # raises lxml.etree.DocumentInvalid on failure
print("Valid.")
```
- For multiple XSD files, point to the “root” XSD that imports the others.

### 4) .NET (C#)
```csharp
using System;
using System.Xml;
using System.Xml.Schema;

class Program {
  static void Main() {
    var settings = new XmlReaderSettings();
    settings.ValidationType = ValidationType.Schema;
    settings.Schemas.Add(null, "sdc-root.xsd");
    settings.ValidationEventHandler += (o, e) =>
      Console.WriteLine($"{e.Severity}: {e.Message}");

    using var reader = XmlReader.Create("my-document.xml", settings);
    while (reader.Read()) { }
    Console.WriteLine("Completed schema validation.");
  }
}
```

## Dealing with namespaces
- In your XML, the default namespace and any prefixes (e.g., `sdc:`) must match the schema’s `targetNamespace`.
- If your document uses multiple namespaces, the root XSD usually **imports** the others. Validate against that root so the validator knows about all namespaces.

## Common errors and how to fix them
- **“Cannot find the declaration of element …”**  
  The root element’s namespace doesn’t match the schema, or you’re pointing at the wrong XSD. Confirm `xmlns` and `targetNamespace` values.
- **Datatype issues (e.g., `xs:dateTime`)**  
  Make sure values follow the XML Schema lexical form (e.g., `YYYY-MM-DDThh:mm:ssZ` for UTC).
- **Unexpected elements/attributes**  
  Your document has fields not allowed by the schema. Remove them or extend the schema as intended by the SDC spec.
- **Missing required elements**  
  Add the required nodes in the correct order if the content model is sequence‑based.
- **Import/include resolution**  
  Keep related XSDs together and avoid changing their filenames unless you update import paths.

## Making validation part of your workflow
- **Local checks** before committing changes.
- **Automated validation** in CI: run `xmllint` (or your validator) and fail the build on errors.
- **Pre-commit hooks** for quick feedback while editing forms or instances.

## Minimal checklist
- [ ] XML declares the expected namespaces.
- [ ] You validate against the **root** SDC XSD.
- [ ] All referenced/imported XSDs are present and reachable.
- [ ] CI job fails on validation errors (not just warnings).
- [ ] Validation output is visible in build logs with line/column numbers.
