# Outline

Levels of validation
* Schema Validation
* is the template a valid CAP template - version, release, other metadata
* Does your data dictionary accurately reflect all the data and metadata that is in the empty CAP template? 
* Authentication and authorization for blank templates and completed form - are you allowed to see the data 
* Validation errors - test cases and validation messages 
* Validate attributes in FDF-R - validate against FDF to check if the rules specified in the FDF are followed 
    * e.g. minCard, maxCard, repeat numbers, readOnly, newData, changedData
* list all the attributes that require extra validation so that code can be written against descriptions to validate

## necessary features
* what are the possible codes in the NAACCR OBR? (can start with tables in the program that updated quarterly)
* Check if there is a repeat - is the integer in the appropriate place in the ID?
* pull out baseURI to identify custom elements from SDC form (in the NAACCR Vol V message would need to validate it against the ID - note that a custom property or element may no be present in the NAACCR Vol V)
* Detect non-eCP OBXs which should be in a separate OBR e.g. a gross report that was part of the eCP OBR, but should be separated
* need the template in order to apprioriately validate that the rule was followed - custom rules which are not part of the template should generate a warning
